```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# create an 1-sphere and see what 2D geometric algebra does to it

def map2d(u0, u1, u2, u12):
    return np.array([
        [u0 + u1, u2 + u12],
        [u2 - u12, u0 - u1]
    ])

# I will now use the convention to describe 2D objects  
```

```python
# get a set of points from a 1-at:sphere, equidistant
n = 100
theta = np.linspace(0, 2 * np.pi, n)
x = np.cos(theta)
y = np.sin(theta)
fig, ax = plt.subplots(figsize = (10, 10))
plt.scatter(x, y, c = theta/np.max(theta))
```

```python
M = np.array((x, y))
# now experiment with the geometric algebra
u_s = np.array((0, 0, 1, 0))
u_s = list(u_s/(u_s.dot(u_s))**(1/2))
plt.scatter(*map2d(*u_s) @ M, c = theta/np.max(theta))
```
